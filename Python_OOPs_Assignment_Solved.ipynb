**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    # getters
    def get_name(self):
        return self.__name

    def get_roll_number(self):
        return self.__roll_number

    # setters
    def set_name(self, name):
        self.__name = name

    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number


s1 = Student("Aftab", 101)
print("Before:", s1.get_name(), s1.get_roll_number())

s1.set_name("Ali")
s1.set_roll_number(202)
print("After:", s1.get_name(), s1.get_roll_number())
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance: {self.__balance}")
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount):
        # balance zero se kam nahi hona chahiye
        if amount > self.__balance:
            print("Insufficient balance!")
        elif amount <= 0:
            print("Withdraw amount must be positive")
        else:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance: {self.__balance}")

    def get_balance(self):
        return self.__balance


acc = BankAccount(1000)
acc.deposit(500)
acc.withdraw(300)
acc.withdraw(5000)   # balance se zyada, reject hoga
print("Final balance:", acc.get_balance())
```

**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

```python
# _protected -> sirf convention hai, access ho sakta hai (subclass ke liye hota hai)
# __private  -> name mangling hoti hai, bahar se seedha access nahi hota

class Parent:
    def __init__(self):
        self._protected_var = "I am protected"
        self.__private_var = "I am private"

    def show(self):
        print(self._protected_var)
        print(self.__private_var)


class Child(Parent):
    def access(self):
        print("Child accessing protected:", self._protected_var)   # works
        try:
            print(self.__private_var)                              # fail
        except AttributeError as e:
            print("Child accessing private -> Error:", e)


p = Parent()
p.show()

print("\nOutside class:")
print(p._protected_var)          # chal jata hai (but not recommended)
try:
    print(p.__private_var)
except AttributeError as e:
    print("Private error:", e)

print("\nInside child class:")
c = Child()
c.access()

# private ko name mangling se access kiya ja sakta hai (not recommended)
print("\nName mangled access:", p._Parent__private_var)
```

**Q5:** Encapsulation ka real-world faida kya hai? Private variable ko class ke bahar se direct access karne par kaunsa error aata hai? Code likh kar check karein.

**Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Employee(Person):
    def __init__(self, name, age, company):
        # super() parent ke __init__ ko call karta hai taake name aur age set ho jayein
        super().__init__(name, age)
        self.company = company

    def show(self):
        print(f"Name: {self.name}, Age: {self.age}, Company: {self.company}")


e = Employee("Aftab", 25, "First Light Technology")
e.show()
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Hierarchical Inheritance: aik parent class se multiple child classes banti hain

import math

class Shape:
    def __init__(self, name):
        self.name = name

    def info(self):
        print(f"This is a {self.name}")


class Circle(Shape):
    def __init__(self, radius):
        super().__init__("Circle")
        self.radius = radius

    def area(self):
        return round(math.pi * self.radius ** 2, 2)


class Square(Shape):
    def __init__(self, side):
        super().__init__("Square")
        self.side = side

    def area(self):
        return self.side ** 2


c = Circle(7)
s = Square(4)

c.info()
print("Circle area:", c.area())
s.info()
print("Square area:", s.area())
```

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

```python
# Method Overriding: child class parent ke method ko apne tareeqe se dobara likhti hai

class Bird:
    def fly(self):
        print("Most birds can fly")


class Ostrich(Bird):
    def fly(self):
        print("Ostriches cannot fly")


b = Bird()
o = Ostrich()

b.fly()
o.fly()
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Operator Overloading: + jaise operator ko apni class ke liye custom behaviour dena

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"


p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2      # p1.__add__(p2) call hota hai
print("p1 + p2 =", p3)
```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Python mein method overloading default nahi hoti. Same name ka method dobara likho
# to aakhri wala purane ko replace kar deta hai.

class Demo:
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):     # upar wala add overwrite ho gaya
        return a + b + c


d = Demo()
try:
    print(d.add(1, 2))
except TypeError as e:
    print("Error:", e)

# Solution: default arguments ya *args use karo
class Calculator:
    def add(self, *args):
        return sum(args)


calc = Calculator()
print(calc.add(1, 2))
print(calc.add(1, 2, 3))
print(calc.add(1, 2, 3, 4, 5))


# Aik aur tareeqa: default arguments
class Calculator2:
    def add(self, a, b=0, c=0):
        return a + b + c


calc2 = Calculator2()
print(calc2.add(5))
print(calc2.add(5, 10))
print(calc2.add(5, 10, 15))
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


class Laptop(Computer):
    def process(self):
        print("Laptop is processing data on battery power")


class Desktop(Computer):
    def process(self):
        print("Desktop is processing data with high performance")


Laptop().process()
Desktop().process()

# Agar child process() implement na kare to wo bhi abstract rahegi
class Tablet(Computer):
    pass

try:
    Tablet()
except TypeError as e:
    print("Error:", e)
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
from abc import ABC, abstractmethod

class Car(ABC):
    # concrete method
    def start_engine(self):
        print("Engine started")

    # abstract method
    @abstractmethod
    def drive(self):
        pass


class Sedan(Car):
    def drive(self):
        print("Sedan is driving smoothly")


my_car = Sedan()
my_car.start_engine()   # parent ka concrete method
my_car.drive()          # child ka implemented method
```

**Q19:** Abstraction aur Encapsulation ke darmiyan kya wazeh farq (main difference) hai? Dono ke concepts ko short statements mei define karein.

```python
# Abstraction:
# Sirf zaroori cheezein dikhana aur implementation details chhupana.
# (Design level - "kya karna hai", "kaise" nahi). Achieve hota hai abstract class se.

# Encapsulation:
# Data aur methods ko aik class mein bundle karna aur data ko private bana kar
# direct access se bachana. (Data protection). Achieve hota hai private variables se.

print("Abstraction  -> hides complexity (what to do)")
print("Encapsulation -> hides/protects data (how data is accessed)")

# Example
# Abstraction: car chalane ke liye sirf steering, brake chahiye, engine ke andar ka kaam nahi dikhta.
# Encapsulation: engine ki settings private hain, sirf allowed methods se change hoti hain.
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    def __init__(self, amount):
        self.amount = amount

    @abstractmethod
    def pay(self):
        pass


class CreditCardPayment(Payment):
    def __init__(self, amount, card_number):
        super().__init__(amount)
        self.card_number = card_number

    def pay(self):
        masked = "**** **** **** " + self.card_number[-4:]
        print(f"Paid {self.amount} using Credit Card {masked}")


class UPIPayment(Payment):
    def __init__(self, amount, upi_id):
        super().__init__(amount)
        self.upi_id = upi_id

    def pay(self):
        print(f"Paid {self.amount} using UPI ID {self.upi_id}")


payments = [
    CreditCardPayment(2500, "1234567812345678"),
    UPIPayment(1200, "aftab@upi"),
]

for p in payments:
    p.pay()
```
